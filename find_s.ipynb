```python
# ============================================================
# FIND-S ALGORITHM
# Dataset: Employee Promotion Prediction
# ============================================================

def find_s(training_data, target_col="Promotion", positive_label="Yes"):

    # Get all attributes except the target column
    attributes = [
        attribute for attribute in training_data[0]
        if attribute != target_col
    ]

    # Start with the most specific hypothesis
    hypothesis = {attribute: "0" for attribute in attributes}

    print("Initial Hypothesis:")
    print(hypothesis)
    print("-" * 50)

    # Process each training example
    for i, example in enumerate(training_data, start=1):

        print(f"\nExample {i}: {example}")

        # FIND-S considers only positive examples
        if example[target_col] == positive_label:

            print("Positive Example -> Updating Hypothesis")

            for attribute in attributes:

                # First positive example
                if hypothesis[attribute] == "0":
                    hypothesis[attribute] = example[attribute]

                # If values differ, generalize
                elif hypothesis[attribute] != example[attribute]:
                    hypothesis[attribute] = "?"

            print("Current Hypothesis:")
            print(hypothesis)

        else:
            print("Negative Example -> Ignored by FIND-S")

    return hypothesis


# ------------------------------------------------------------
# Dataset: Employee Promotion Prediction
# Total Records = 20
# ------------------------------------------------------------

training_data = [

    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "High", "Promotion": "Yes"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "Medium", "Promotion": "Yes"},
    {"Performance": "Average", "Experience": "Low", "Training": "Incomplete", "Attendance": "Low", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "Low", "Promotion": "Yes"},
    {"Performance": "Good", "Experience": "Medium", "Training": "Completed", "Attendance": "High", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "High", "Promotion": "Yes"},
    {"Performance": "Poor", "Experience": "Low", "Training": "Incomplete", "Attendance": "Low", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "Medium", "Promotion": "Yes"},
    {"Performance": "Average", "Experience": "Medium", "Training": "Incomplete", "Attendance": "Medium", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "High", "Promotion": "Yes"},
    {"Performance": "Good", "Experience": "Low", "Training": "Incomplete", "Attendance": "Low", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "Low", "Promotion": "Yes"},
    {"Performance": "Average", "Experience": "Medium", "Training": "Incomplete", "Attendance": "High", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "Medium", "Promotion": "Yes"},
    {"Performance": "Poor", "Experience": "Low", "Training": "Completed", "Attendance": "Low", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "High", "Promotion": "Yes"},
    {"Performance": "Good", "Experience": "Medium", "Training": "Incomplete", "Attendance": "Low", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "Low", "Promotion": "Yes"},
    {"Performance": "Average", "Experience": "Low", "Training": "Incomplete", "Attendance": "Medium", "Promotion": "No"},
    {"Performance": "Excellent", "Experience": "High", "Training": "Completed", "Attendance": "Medium", "Promotion": "Yes"}

]


# ------------------------------------------------------------
# Run the FIND-S Algorithm
# ------------------------------------------------------------

final_hypothesis = find_s(training_data)


print("\n" + "=" * 50)
print("FINAL HYPOTHESIS")
print("=" * 50)
print(final_hypothesis)
```

```python

# ============================================================
# DATA VISUALIZATION
# Dataset: Employee Promotion Prediction
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Convert existing training_data into DataFrame
df = pd.DataFrame(training_data)


# ------------------------------------------------------------
# GRAPH 1: Promotion Distribution
# ------------------------------------------------------------

promotion_count = df["Promotion"].value_counts()

plt.figure(figsize=(6, 4))

plt.bar(
    promotion_count.index,
    promotion_count.values
)

plt.title("Employee Promotion Distribution")
plt.xlabel("Promotion Status")
plt.ylabel("Number of Employees")

plt.show()


# ------------------------------------------------------------
# GRAPH 2: Performance vs Promotion
# ------------------------------------------------------------

performance_graph = pd.crosstab(
    df["Performance"],
    df["Promotion"]
)

performance_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Performance vs Promotion")
plt.xlabel("Performance")
plt.ylabel("Number of Employees")

plt.xticks(rotation=0)
plt.legend(title="Promotion")

plt.show()


# ------------------------------------------------------------
# GRAPH 3: Experience vs Promotion
# ------------------------------------------------------------

experience_graph = pd.crosstab(
    df["Experience"],
    df["Promotion"]
)

experience_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Experience vs Promotion")
plt.xlabel("Experience Level")
plt.ylabel("Number of Employees")

plt.xticks(rotation=0)
plt.legend(title="Promotion")

plt.show()


# ------------------------------------------------------------
# GRAPH 4: Training vs Promotion
# ------------------------------------------------------------

training_graph = pd.crosstab(
    df["Training"],
    df["Promotion"]
)

training_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Training vs Promotion")
plt.xlabel("Training Status")
plt.ylabel("Number of Employees")

plt.xticks(rotation=0)
plt.legend(title="Promotion")

plt.show()


# ------------------------------------------------------------
# GRAPH 5: Attendance vs Promotion
# ------------------------------------------------------------

attendance_graph = pd.crosstab(
    df["Attendance"],
    df["Promotion"]
)

attendance_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Attendance vs Promotion")
plt.xlabel("Attendance Level")
plt.ylabel("Number of Employees")

plt.xticks(rotation=0)
plt.legend(title="Promotion")

plt.show()
```
